```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv('/content/Complete_Techno_Sales_Data-2.csv')
df[['Sales', 'Quantity']].hist(figsize=(10, 4), bins=15)
plt.suptitle("Sales and Quantity Distributions")
plt.tight_layout()
plt.show()
plt.figure(figsize=(8, 4))
sns.boxplot(data=df[['Cost', 'Sales']])
plt.title("Cost and Sales Range & Outliers")
plt.show()
num_cols = df.select_dtypes(include=['number']).columns
if len(num_cols) > 1:
    plt.figure(figsize=(6, 4))
    sns.heatmap(df[num_cols].corr(), annot=True, cmap='coolwarm', fmt=".2f")
    plt.title("Correlation Analysis Heatmap")
    plt.show()
if len(num_cols) >= 2:
    plt.figure(figsize=(7, 5))
    sns.scatterplot(x=num_cols[-2], y=num_cols[-1], data=df, hue='Category' if 'Category' in df.columns else None)
    plt.title(f"{num_cols[-2]} vs {num_cols[-1]} Analysis")
    plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
    plt.tight_layout()
    plt.show()
```
